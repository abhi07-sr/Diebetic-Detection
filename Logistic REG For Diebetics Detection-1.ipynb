```python
df['Pregnancies'] = df["Pregnancies"].replace(0, df['Pregnancies'].median())
df['Glucose'] = df["Glucose"].replace(0, df['Glucose'].mean())
df['BloodPressure'] = df["BloodPressure"].replace(0, df['BloodPressure'].mean())
df['SkinThickness'] = df["SkinThickness"].replace(0, df['SkinThickness'].median())
df['Insulin'] = df["Insulin"].replace(0, df['Insulin'].median())
df['BMI'] = df["BMI"].replace(0, df['BMI'].mean())
df["Age"] = df["Age"].replace(0, df['Age'].median())
df.isnull().sum()
```

```python
fig,ax =plt.subplots(figsize=(15,15))
sns.boxplot(data=df, ax=ax)
plt.savefig('boxplot.png', dpi=500, bbox_inches='tight')
```

```python
x = df.drop("Outcome", axis=1)
y = df["Outcome"]
```

```python
cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
mask = pd.Series(True, index=x.index)
for col in cols:
    q1 = x[col].quantile(0.25)
    q3 = x[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    mask &= (x[col] >= lower_bound) & (x[col] <= upper_bound)
```

```python
x_outlier_detected = x[mask].reset_index(drop=True)
y_outlier_detected = y[mask].reset_index(drop=True)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x = scaler.fit_transform(x_outlier_detected)
```

```python
x_scaled = pd.DataFrame(x, columns=x_outlier_detected.columns)
```
